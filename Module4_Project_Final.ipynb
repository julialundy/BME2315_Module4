## Primary Question: 

Is sustained angiogenesis predictive of early stage (I-II) vs. late stage (III–IV) clear cell renal cell carcinoma? 

## Disease Background: 

Clear cell renal cell carcinoma (ccRCC) is the most common subtype of kidney cancer: it makes up roughly 70% of cases [1]. It is projected that 80,450 Americans will develop kidney cancer in 2026 [2]. ccRCC begins in epithelial cells lining the nephron's proximal tubules [3]. Its main driver is loss of VHL, a tumor suppressor that typically marks hypoxia-inducible factors (HIFs) for degradation [3]. Without VHL, HIFs accumulate and increase vascular endothelial growth factor (VEGF) production, which prompts angiogenesis, the creation of tumor-supporting blood vessels [3]. Mutations in PBRM1, ARID1A, and SMARCA4, parts of a chromatin remodeling complex, alter which genes get expressed [4]. Many tumors are found through imaging [5], but symptoms can include blood in the urine, one-sided lower back pain, and a mass in the side [6]. Surgery is used for localized tumors, while advanced cases are treated with immunotherapy, VEGF-targeting drugs, and HIF-2α inhibitors [7]. Currently, CD70-targeted CAR T-cells are being evaluated through clinical trials [8]. Importantly, mortality increases with stage: 5-year survival is 93.6% for localized disease but only 20.3% when the cancer has disseminated to distant organs [9]. Hence, this project asks if sustained angiogenesis predicts early (I–II) vs. late (III–IV) stage ccRCC.
